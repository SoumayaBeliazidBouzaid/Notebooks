Les *arguments nommés* sont très pratiques pour définir des fonctions à nombre d'arguments variable, surtout quand les valeurs par défaut conviennent à la plupart des appels. En physique, c'est idéal pour des paramètres d'expérience :

````python
def position(t, x0=0.0, v0=0.0, a=9.81):
    """Position d'un mobile en mouvement uniformément accéléré (unités SI)."""
    return x0 + v0 * t + 0.5 * a * t**2

print(position(2.0))                 # chute libre depuis x0 = 0, v0 = 0
print(position(2.0, v0=5.0))         # avec une vitesse initiale
print(position(2.0, x0=10, a=0.0))   # mouvement uniforme
````

## Passage par valeur

Peut-on modifier la valeur d'une variable à l'intérieur d'une fonction ? La plupart des langages (C, Java…) distinguent le « passage par valeur » du « passage par référence ». En Python, cette distinction est un peu artificielle, et il est parfois subtil de savoir si vos variables seront modifiées ou non. Heureusement, il existe des règles claires.

Les paramètres des fonctions sont des références à des objets, **passées par valeur**. Quand vous passez une variable à une fonction, Python passe la référence à l'objet auquel la variable se réfère (la **valeur**), et non la variable elle-même.

Si la **valeur** passée est immuable, la fonction ne modifie pas la variable de l'appelant. Si la **valeur** est modifiable, la fonction peut modifier la variable de l'appelant sur place :

````python
def essaie_de_modifier(x, y, z):
    x = 23
    y.append(42)
    z = [99] # nouvelle référence
    print(x)
    print(y)
    print(z)
````

````python
a = 77    # variable immuable
b = [99]  # variable modifiable
c = [28]
essaie_de_modifier(a, b, c)
````

````python
print(a)
````

````python
print(b)
````

````python
print(c)
````

Les fonctions ont une table de variables locales appelée *espace de noms local*. La variable `x` n'existe qu'à l'intérieur de la fonction `essaie_de_modifier`.

## Les fonctions sont des objets

Les fonctions sont des objets de première classe : elles peuvent être

Ce dernier point est très utile en physique : on peut écrire une fonction générique qui reçoit une *autre fonction* en argument, par exemple pour calculer une dérivée numérique :

````python
def derivee(f, x, h=1e-6):
    """Dérivée numérique de f en x (différence centrée)."""
    return (f(x + h) - f(x - h)) / (2 * h)

def carre(x):
    return x**2

derivee(carre, 3.0)   # on attend 6
````

> **Exercice : suite de Fibonacci**
>
>
> Écrivez une fonction qui affiche les `n` premiers termes de la suite de Fibonacci, définie par :
>
> $$
> \begin{align}
> U_{0} &= 0 \\
> U_{1} &= 1 \\
> U_{n+2} &= U_{n+1} + U_{n}
> \end{align}
> $$

````python
# Votre code ici

def fibonacci(n):
    a = 0
    b = 1

    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b

# Exemple
n = int(input("Donner le nombre de termes : "))
fibonacci(n)
````

````python
#@title Solution { display-mode: "form" }
def fib(n):
    """Affiche les n premiers termes de la suite de Fibonacci."""
    a, b = 0, 1
    i = 0
    while i < n:
        print(b)
        a, b = b, a+b
        i += 1

fib(10)
````

> **Exercice : tri rapide (*Quicksort*)**
>
>
> Implémentez l'[algorithme de tri rapide, tel que défini par Wikipédia](https://fr.wikipedia.org/wiki/Tri_rapide) :
>
> ```text
> fonction quicksort(tableau)
>     var liste inferieurs, superieurs
>     si longueur(tableau) < 2
>         renvoyer tableau
>     choisir et retirer une valeur pivot du tableau
>     pour chaque x du tableau
>         si x < pivot alors ajouter x à inferieurs
>         sinon ajouter x à superieurs
>     renvoyer concaténation(quicksort(inferieurs), pivot, quicksort(superieurs))
> ```

````python
# Votre code ici

def quicksort(tableau):
    inferieurs = []
    superieurs = []

    # Cas de base
    if len(tableau) < 2:
        return tableau

    # Choisir le pivot
    pivot = tableau[0]

    # Classer les éléments
    for x in tableau[1:]:
        if x < pivot:
            inferieurs.append(x)
        else:
            superieurs.append(x)

    # Trier récursivement et concaténer
    return quicksort(inferieurs) + [pivot] + quicksort(superieurs)


# Exemple d'utilisation
tableau = [8, 3, 1, 7, 4, 9, 2, 6]
print("Tableau initial :", tableau)
print("Tableau trié :", quicksort(tableau))
````

````python
#@title Solution { display-mode: "form" }
def qsort(lst):
    """Tri rapide : renvoie une copie triée de la liste."""
    if len(lst) <= 1:
        return lst
    pivot, reste = lst[0], lst[1:]

    inferieurs = [x for x in reste if x < pivot]
    superieurs = [x for x in reste if x >= pivot]
    return qsort(inferieurs) + [pivot] + qsort(superieurs)

# Vérifions que qsort trie bien :
assert qsort(list(range(10))) == list(range(10))
assert qsort(list(range(10))[::-1]) == list(range(10))
assert qsort([1, 4, 2, 5, 3]) == sorted([1, 4, 2, 5, 3])
print("Tests réussis")
````

## Intercepter les exceptions

### try / except

````python
for texte in ['a', '3.5', '42']:
    try:
        x = int(texte)
        print(texte, '-> entier', x)
    except ValueError:
        print(texte, "-> ce n'est pas un entier valide, on continue...")
````

(Avec `input()`, on peut demander une saisie à l'utilisateur : `x = int(input('Entrez un nombre : '))`.)

### try / finally

````python
try:
    x = int('a')
finally:
    print('Merci pour votre saisie')
````

Le bloc `finally` est toujours exécuté, même en cas d'erreur (ici, l'erreur est ensuite affichée). C'est important pour la gestion des ressources (par exemple pour fermer un fichier).

### Plus facile de demander pardon que la permission

````python
def affiche_trie(collection):
    try:
        collection.sort()
    except AttributeError:
        pass # L'instruction pass ne fait rien
    print(collection)
````

````python
affiche_trie([1, 3, 2])
````

````python
affiche_trie(set((1, 3, 2)))
````

````python
affiche_trie('132')
````

## Lever des exceptions

### Capturer et relancer une exception

````python
def filtre_nom(nom):
    try:
        nom = nom.encode('ascii')
    except UnicodeError as e:
        if nom == 'Gaël':
            print('OK, Gaël')
        else:
            raise e
    return nom

filtre_nom('Gaël')
````

````python
# Erreur volontaire
filtre_nom('Stéfan')
````

### Des exceptions pour transmettre des messages entre parties du code

````python
def fleche_achille(x):
    if abs(x - 1) < 1e-3:
        raise StopIteration
    x = 1 - (1-x)/2.
    return x

x = 0

while True:
    try:
        x = fleche_achille(x)
    except StopIteration:
        break

x
````

Utilisez les exceptions pour signaler que certaines conditions sont remplies (par exemple `StopIteration`) ou non (levée d'une erreur personnalisée).

````python
# Votre code ici

def vitesse(distance, duree):
    if duree <= 0:
        raise ValueError("La durée doit être strictement positive.")

    return distance / duree


# Test avec try/except
try:
    d = float(input("Donner la distance (en mètres) : "))
    t = float(input("Donner la durée (en secondes) : "))

    v = vitesse(d, t)
    print("La vitesse moyenne est :", v, "m/s")

except ValueError as e:
    print("Erreur :", e)
````

````python
#@title Solution { display-mode: "form" }
def vitesse(distance, duree):
    """Vitesse moyenne (m/s) pour une distance (m) et une durée (s)."""
    if duree <= 0:
        raise ValueError("La durée doit être strictement positive.")
    return distance / duree

for d, t in [(100, 9.58), (100, 0)]:
    try:
        print(f"v = {vitesse(d, t):.2f} m/s")
    except ValueError as e:
        print("Erreur :", e)
````
